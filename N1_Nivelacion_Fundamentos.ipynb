```python
# ============================================================
# BLOQUE 2 - CONDICIONALES
# ============================================================
print("========== BLOQUE 2: CONDICIONALES ==========")

print("\n===== ESTADO DEL PERSONAJE =====")
vida     = 25.0
vida_max = 100.0
pct      = (vida / vida_max) * 100

if pct <= 0:
    estado = 'MUERTO'
elif pct <= 25:
    estado = 'CRITICO'
elif pct <= 50:
    estado = 'HERIDO'
elif pct <= 75:
    estado = 'ESTABLE'
else:
    estado = 'SALUDABLE'

print(f'Vida: {vida:.0f}/{vida_max:.0f} ({pct:.0f}%) -> Estado: {estado}')

print("\n===== TIPO DE ATAQUE =====")
clase_personaje = 'Mago'
nivel_habilidad = 3

match clase_personaje:
    case 'Guerrero': tipo_ataque = 'Espada'
    case 'Mago':     tipo_ataque = 'Hechizo'
    case 'Arquero':  tipo_ataque = 'Flecha'
    case _:          tipo_ataque = 'Puno'

puede_usar_magia = (clase_personaje == 'Mago' and nivel_habilidad >= 3)

if puede_usar_magia:
    print(f'Clase: {clase_personaje} -> Bola de fuego!')
else:
    print(f'Clase: {clase_personaje} -> {tipo_ataque} basico')

print("\n===== EJERCICIO 2A: COMBATE =====")
vida_enemigo   = 40
ataque_jugador = 35
nivel_jugador  = 6

bonificacion  = 10 if nivel_jugador >= 5 else 0
dano_total    = ataque_jugador + bonificacion
vida_restante = vida_enemigo - dano_total

print(f'Ataque base:   {ataque_jugador}')
print(f'Bonificacion:  {bonificacion}')
print(f'Dano total:    {dano_total}')
print(f'Vida restante: {vida_restante}')

if vida_restante <= 0:
    print('Resultado: Enemigo derrotado! +50 XP')
elif vida_restante <= 20:
    print('Resultado: Enemigo en estado critico')
else:
    print(f'Resultado: Enemigo resiste. Vida restante: {vida_restante}')
```

```python
# ============================================================
# BLOQUE 3 - CICLOS
# ============================================================
print("========== BLOQUE 3: CICLOS ==========")

print("\n===== INVENTARIO =====")
inventario = [
    'Espada de hierro',
    'Pocion de vida',
    'Escudo de madera',
    'Llave dorada'
]

print("-- Lista numerada --")
for i, item in enumerate(inventario, 1):
    print(f'{i}. {item}')

buscar = 'Pocion de vida'
if buscar in inventario:
    print(f'[OK] {buscar} encontrada')
else:
    print(f'[X]  {buscar} no disponible')

print("\n===== COMBATE POR RONDAS =====")
vida_hero    = 80
vida_enemigo = 60
ronda        = 1

while vida_hero > 0 and vida_enemigo > 0:
    vida_enemigo -= 15
    vida_hero    -= 10
    print(f'Ronda {ronda}: Hero={vida_hero} | Enemigo={vida_enemigo}')
    ronda += 1

print('VICTORIA!' if vida_hero > 0 else 'DERROTA')

print("\n===== SUBIR NIVEL =====")
xp           = 0
nivel        = 1
xp_necesario = 100
batallas     = [20, 35, 15, 40, 30]

for xp_ganado in batallas:
    xp += xp_ganado
    print(f'XP acumulado: {xp}')
    if xp >= xp_necesario:
        nivel += 1
        xp -= xp_necesario
        print(f'*** Subiste al Nivel {nivel}! XP restante: {xp} ***')
```

```python
# ============================================================
# BLOQUE 4 - FUNCIONES
# ============================================================
print("========== BLOQUE 4: FUNCIONES ==========")

def calcular_dano(ataque: int, defensa: int) -> int:
    '''Retorna el dano real (minimo 1)'''
    dano = ataque - defensa
    return dano if dano > 0 else 1

def aplicar_curacion(vida: float, cur: float, max_vida: float) -> float:
    '''Cura sin pasar el maximo'''
    nueva = vida + cur
    return min(nueva, max_vida)

def mostrar_estado(nombre: str, vida: float, nivel: int):
    '''Imprime el estado del personaje'''
    print(f'{nombre} [Nv{nivel}] HP: {vida:.0f}')

def subir_nivel(xp_actual: int, xp_necesario: int, nivel_actual: int) -> int:
    '''Sube de nivel si tiene suficiente XP'''
    if xp_actual >= xp_necesario:
        nivel_actual += 1
        print(f'*** Subiste al Nivel {nivel_actual}! ***')
    else:
        print(f'XP insuficiente. Necesitas {xp_necesario - xp_actual} XP mas')
    return nivel_actual

print("\n===== PRUEBA: calcular_dano =====")
d = calcular_dano(20, 8)
print(f'Dano (ataque=20, defensa=8): {d}')
d2 = calcular_dano(5, 10)
print(f'Dano (ataque=5, defensa=10): {d2}')

print("\n===== PRUEBA: aplicar_curacion =====")
v = aplicar_curacion(40, 80, 100)
print(f'Vida tras curacion (40+80, max 100): {v}')
v2 = aplicar_curacion(60, 20, 100)
print(f'Vida tras curacion (60+20, max 100): {v2}')

print("\n===== PRUEBA: mostrar_estado =====")
mostrar_estado('Frodo', 100.0, 3)
mostrar_estado('Aragorn', 75.0, 5)

print("\n===== PRUEBA: subir_nivel =====")
resultado = subir_nivel(110, 100, 3)
print(f'Nivel resultante: {resultado}')
resultado2 = subir_nivel(80, 100, 3)
print(f'Nivel resultante: {resultado2}')
```

```python
# ============================================================
# BLOQUE 5 - EJERCICIO INTEGRADOR FINAL
# ============================================================
print("========== BLOQUE 5: SIMULADOR RPG ==========")

def calcular_dano(ataque: int, defensa: int) -> int:
    dano = ataque - defensa
    return dano if dano > 0 else 1

def aplicar_curacion(vida: float, cur: float, max_vida: float) -> float:
    return min(vida + cur, max_vida)

def mostrar_estado(nombre: str, vida: float, nivel: int):
    print(f'  {nombre} [Nv{nivel}] HP: {vida:.0f}')

def combatir(vida_hero, ataque_hero, def_hero, vida_enemigo, ataque_enemigo, nombre_enemigo):
    print(f'\n Combate contra {nombre_enemigo}!')
    ronda = 1
    while vida_hero > 0 and vida_enemigo > 0:
        dano_hero     = calcular_dano(ataque_hero, 0)
        vida_enemigo -= dano_hero
        dano_enemigo  = calcular_dano(ataque_enemigo, def_hero)
        vida_hero    -= dano_enemigo
        print(f'  Ronda {ronda}: Hero HP={max(vida_hero,0):.0f} | '
              f'{nombre_enemigo} HP={max(vida_enemigo,0):.0f}')
        ronda += 1
    if vida_hero > 0:
        print(f'  -> {nombre_enemigo} derrotado!')
    else:
        print(f'  -> El heroe murio...')
    return vida_hero

nombre_hero  = 'Guerrero'
nivel_hero   = 2
vida_hero    = 80.0
vida_max     = 80.0
ataque_hero  = 18
defensa_hero = 8

enemigos = [
    {'nombre': 'Goblin', 'vida': 40,  'ataque': 8},
    {'nombre': 'Orco',   'vida': 70,  'ataque': 14},
    {'nombre': 'Dragon', 'vida': 120, 'ataque': 25}
]

print(f'\nHEROE: {nombre_hero} Nv.{nivel_hero} | HP: {vida_hero} | ATK: {ataque_hero} | DEF: {defensa_hero}')
print('='*50)

enemigos_derrotados = 0

for enemigo in enemigos:
    if vida_hero <= 0:
        break
    vida_hero = combatir(
        vida_hero, ataque_hero, defensa_hero,
        enemigo['vida'], enemigo['ataque'], enemigo['nombre']
    )
    if vida_hero > 0:
        enemigos_derrotados += 1
        vida_hero = aplicar_curacion(vida_hero, 20, vida_max)
        print(f'  -> Curacion aplicada! HP actual: {vida_hero:.0f}')

print('\n' + '='*50)
print('RESULTADO FINAL:')
print(f'  Enemigos derrotados: {enemigos_derrotados}/3')
print(f'  Vida restante:       {max(vida_hero, 0):.0f}')

if enemigos_derrotados == 3:
    print('  VICTORIA TOTAL!')
elif enemigos_derrotados > 0:
    print(f'  Caiste en combate tras {enemigos_derrotados} victorias.')
else:
    print('  DERROTA. El heroe murio en el primer combate.')
```
