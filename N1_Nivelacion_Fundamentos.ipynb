```python
# ============================================================
# BLOQUE 1 - TIPOS DE DATOS Y VARIABLES
# ============================================================
print("========== BLOQUE 1: TIPOS DE DATOS ==========")

# ── Variables del personaje Aragorn ──
nombre      = 'Aragorn'
nivel       = 1
vida        = 100.0
vida_maxima = 100.0
esta_vivo   = True
clase       = 'Guerrero'
puntos_ataque  = 15
puntos_defensa = 10

# Mostrar tipos de cada variable
print(type(nombre))      # <class 'str'>
print(type(nivel))       # <class 'int'>
print(type(vida))        # <class 'float'>
print(type(esta_vivo))   # <class 'bool'>

# Mostrar datos de Aragorn
print(f'{nombre} (Nv.{nivel}) - Vida: {vida}')

# ── Ejercicio 1A - Mago Gandalf ──
print("\n===== EJERCICIO 1A: GANDALF =====")
nombre_mago = 'Gandalf'
nivel_mago  = 5
vida_mago   = 80.0
clase_mago  = 'Mago'
mana        = 120

print(f'{nombre_mago} [{clase_mago}] Nv.{nivel_mago} | Vida: {vida_mago} | Mana: {mana}')

# ── Casting (conversion de tipos) ──
print("\n===== CASTING =====")
ataque = 15
dano   = float(ataque) * 1.5    # int -> float
msg    = 'Dano: ' + str(dano)   # float -> str
print(msg)

# Leer datos del usuario
print("\n===== INGRESO DE DATOS =====")
nom = input('Ingresa tu nombre de heroe: ')
niv = int(input('Ingresa tu nivel: '))
print(f'Bienvenido: {nom} Nv.{niv}')

# Formato especial con f-string
print(f'Vida actual: {vida:.1f}%')
```
